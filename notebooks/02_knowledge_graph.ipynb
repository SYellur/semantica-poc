# 2. Build the knowledge graph

This notebook follows the existing runtime path:

`load_data()` → `source_to_kg()` → `GraphBuilder().build()`

The adapter is Northstar-specific. Semantica's `GraphBuilder` builds the graph from the entities and relationships it receives.

## Load and map the data

```python
from northstar_authz.service import load_data
from northstar_authz.adapter import source_to_kg

data = load_data()
graph_input = source_to_kg(data)

print("Entities:", len(graph_input["entities"]))
print("Relationships:", len(graph_input["relationships"]))
```

## Look at a mapped relationship

```python
next(
    edge for edge in graph_input["relationships"]
    if edge["type"] == "memberOf" and edge["source"] == "Alice"
)
```

## Build the graph with Semantica

```python
from semantica.kg import GraphBuilder

graph = GraphBuilder().build(graph_input)
print("Graph keys:", graph.keys())
print("Relationship count:", graph.get("num_relationships"))
```

## Use the project's service helper

`build_graph` in `northstar_authz/service.py` wraps the same Semantica call. Prefer this helper in application code so the graph-building step stays in one place.

```python
from northstar_authz.service import build_graph

graph_from_helper = build_graph(data)
print("Relationship count:", graph_from_helper.get("num_relationships"))
```

**Key point:** `adapter.py` maps Northstar source fields to the graph relationship names used by this POC, such as `memberOf`, `hasParentTeam`, `directShareTo`, and `restrictedFrom`. It is mapping code, not a reasoning engine.

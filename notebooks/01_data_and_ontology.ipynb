# 1. Explore the Northstar data and ontology

This notebook introduces the sample data and the curated ontology.

It uses the existing project files; it does not create a new ontology or change the source data.

## Load the sample data

`load_data` is defined in `northstar_authz/service.py`. It reads `data/northstar.json`.

```python
from northstar_authz.service import load_data

data = load_data()
print("People:", len(data["people"]))
print("Teams:", len(data["teams"]))
print("Projects:", len(data["projects"]))
print("Documents:", len(data["documents"]))
```

## Read the curated ontology

The ontology is stored in `ontology/northstar.ttl`. It defines core classes such as `Person`, `Team`, `Project`, and `Document`, plus association classes such as `DirectDocumentShare` and `DocumentRestriction`.

```python
from pathlib import Path
from rdflib import Graph

ontology = Graph()
ontology.parse(Path("ontology/northstar.ttl"), format="turtle")
print("Ontology triples:", len(ontology))
```
